## 3. Notebook Memory and Execution Order

A notebook remembers the code you have already run during the current session.

Run the next two cells **in order**.

````python
favorite_snack = "chips"
print("My favorite snack is", favorite_snack)
````

````python
favorite_snack = "cookies"
print("Now my favorite snack is", favorite_snack)
````

Now run the **first** cell again.

What happened?

The value changed back because the notebook follows the order in which cells are **run**, not simply the order in which they appear on the page.

> The notebook has memory, but it does not have judgment. It remembers whatever you ran most recently.

## A Very Brief Introduction to Data Science 📊

**Data Science** is the process of using data to answer questions, discover patterns, and make informed decisions.

A basic analytics workflow usually looks like this:

1. **Collect** the data
2. **Inspect** the data
3. **Analyze** the data
4. **Communicate** what you discovered

In this example, we will analyze fictional student practice data to answer two questions:

- What was the average quiz score?
- Which student earned the highest score?

The data has been collected. Now it is time to politely interrogate it.

````python
import pandas as pd

# 1. COLLECT: Create a small dataset
student_data = {
    "Student": ["Jordan", "Taylor", "Morgan", "Casey", "Avery"],
    "Practice_Minutes": [30, 45, 20, 60, 40],
    "Quiz_Score": [78, 88, 72, 95, 85]
}

data = pd.DataFrame(student_data)

# 2. INSPECT: View the data
print("Student Practice Data:")
display(data)

# 3. ANALYZE: Calculate a summary and find the highest score
average_score = data["Quiz_Score"].mean()
highest_score_row = data.loc[data["Quiz_Score"].idxmax()]

# 4. COMMUNICATE: Share what we discovered
print(f"The average quiz score was {average_score:.1f}.")
print(
    f"{highest_score_row['Student']} earned the highest score "
    f"with {highest_score_row['Quiz_Score']} points."
)
````
